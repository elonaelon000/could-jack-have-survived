**Research question:** Among real Titanic passengers resembling the fictional Jack Dawson, which characteristics were associated with better observed survival?

## 1. Load and profile the data

Only inspect dimensions, column types, column meanings and missingness at this stage. **Do not inspect survival rates by group until Section 2 is recorded in Git history.**


```python
from pathlib import Path
import pandas as pd

data_path = next((p for p in (Path('data/train.csv'), Path('../data/train.csv')) if p.exists()), None)
if data_path is None:
    raise FileNotFoundError('Place the Kaggle Titanic train.csv under data/train.csv')
df = pd.read_csv(data_path)
print('Shape:', df.shape)
display(pd.DataFrame({'dtype': df.dtypes.astype(str), 'missing': df.isna().sum()}))
```

**Dataset:** Kaggle Titanic `train.csv` (891 rows, 12 columns). `Survived` is the observed 0/1 outcome; `Pclass` is ticket class; `Sex` and `Age` are recorded demographics; `SibSp` counts siblings/spouses aboard; `Parch` counts parents/children aboard; `Fare` is recorded fare; `Cabin` and `Embarked` are recorded cabin and embarkation. `PassengerId`, `Name`, and `Ticket` are identifiers/descriptors. Missingness: Age 177; Cabin 687; Embarked 2. These profile facts do not reveal subgroup outcomes.


## 2. Three hypotheses — registered BEFORE outcome exploration

**Jack-like profile:** male, third class, ages **18–25 inclusive**, known age. This is an analytic proxy, not a record of the fictional Jack. Alternative age ranges can be used for labelled robustness checks but must not silently replace the primary definition.

**H1 · Sex (directional):** Among passengers aged **18–50 inclusive**, women have higher observed survival rates than men **within each passenger class**. **Test:** compare female vs male within Pclass=1,2,3 and report both denominators and percentage-point differences.

**H2 · Class (directional):** Among **men aged 18–30 inclusive**, first-class passengers have higher observed survival rates than third-class passengers. **Test:** compare Pclass 1 vs 3 within this restricted age/sex group; report counts, rates and a percentage-point gap.

**H3 · Recorded relatives (directional):** Among **third-class men aged 18–50 inclusive**, those with `SibSp + Parch > 0` have a higher observed survival rate than those with `SibSp + Parch == 0`. **Test:** compare two groups with counts and rates. A zero does **not** prove a passenger travelled alone.

**Evaluation rule:** Hypotheses will be marked SURVIVED or NOT SUPPORTED after analysis; small groups, uncertainty, and competing explanations must be disclosed. Association is not causation. The original wording will not be revised after inspecting results.
